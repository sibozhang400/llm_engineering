```python
# TODO: understand the following code, what do they parse from the website.

# Extract clean text from HTML for LLM processing
from bs4 import BeautifulSoup

soup = BeautifulSoup(website_contents, 'html.parser')

# Remove script, style, and other irrelevant elements
for element in soup(['script', 'style', 'noscript', 'meta', 'link']):
    element.decompose()

# Get text content
clean_text = soup.get_text(separator='\n', strip=True)

# Show a sample
print(f"Clean text length: {len(clean_text)} characters")
print(f"\nFirst 2000 characters of clean text:")
print("="*60)
print(clean_text[:4000])
```
